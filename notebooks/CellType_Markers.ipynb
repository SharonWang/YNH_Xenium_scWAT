```r
Wang_Science2025_top50 = read.csv("/dssg/home/acct-svetoslav_chakarov/svetoslav_chakarov/Data/External_Data/Wang_Science2025_mouse_human/analysis_mouse/Wang_Science2025_mouse_adipose_top50_DE_genes.csv")
Wang_Science2025_top50 <- Wang_Science2025_top50 %>%
  group_by(group) %>%
  mutate(DE_rank = row_number()) %>%
  ungroup()
```

```r
Xenium_panel = read.csv("/dssg/home/acct-svetoslav_chakarov/svetoslav_chakarov/Lab_members/Yanan_Hu/YNH_Xenium/adipose_analysis/Xenium_final_479_gene_annotation_table.csv")
```

```r
library(dplyr)

Wang_Science2025_top50_Xenium <- Wang_Science2025_top50 %>%
  inner_join(
    Xenium_panel,
    by = c("names" = "gene")
  ) %>%
  arrange(group, DE_rank)
```

```r
dim(Wang_Science2025_top50_Xenium)
```

```r
marker_coverage <- Wang_Science2025_top50 %>%
  group_by(group) %>%
  summarise(
    n_top50 = n(),
    n_in_Xenium = sum(names %in% Xenium_panel$gene),
    fraction_in_Xenium = n_in_Xenium / n_top50,
    .groups = "drop"
  ) %>%
  arrange(desc(fraction_in_Xenium))

marker_coverage
```

```r
Wang_Xenium_marker_list <- Wang_Science2025_top50_Xenium %>%
  group_by(group) %>%
  summarise(
    markers = paste(names, collapse = ", "),
    n_markers = n(),
    .groups = "drop"
  )
```
